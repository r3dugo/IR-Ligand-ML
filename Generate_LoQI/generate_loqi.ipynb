## Generating LoQI for CN and NN Ligands

```python
from pathlib import Path
import subprocess
import sys
import os

"""
Note: this notebook assumes structure:
PROJECT_ROOT/
└── Generate_LoQI/
    ├── generate_loqi_generic.py
    └── LoQI/
        ├── scripts/sample_conformers.py
        ├── scripts/conf/loqi/loqi.yaml
        └── data/loqi.ckpt
"""

# Project paths: modify to your actual path if different
PROJECT_ROOT = Path("/Volumes/T7/ML/IR_Ligand_ML")

# Input/output paths
cn_csv = Path(PROJECT_ROOT, "input", "ML-CN-Smiles.csv")
nn_csv = Path(PROJECT_ROOT, "input", "ML-NN-Smiles.csv")
output_base = Path(PROJECT_ROOT, "Generate_LoQI", "output")

# LoQI generation parameters
num_conformers = 12
batch_size = 50

# Should remain the same if the structure is maintained
generate_loqi_dir = Path(PROJECT_ROOT, "Generate_LoQI")
script_path = generate_loqi_dir / "generate_loqi_generic.py"

# LoQI repo paths
loqi_root = generate_loqi_dir / "LoQI"

cmd = [
    sys.executable, str(script_path),
    "--csvs", str(cn_csv), str(nn_csv),
    "--output-base", str(output_base),
    "--n-conformers", str(num_conformers),
    "--batch-size", str(batch_size),
    "--sample-script", str(loqi_root / "scripts" / "sample_conformers.py"),
    "--config", str(loqi_root / "scripts" / "conf" / "loqi" / "loqi.yaml"),
    "--ckpt", str(loqi_root / "data" / "loqi.ckpt"),
]

env = os.environ.copy()
env["KMP_DUPLICATE_LIB_OK"] = "TRUE"

print("Running:")
print(" ".join(cmd))

res = subprocess.run(
    cmd,
    cwd=str(generate_loqi_dir),
    env=env,
    capture_output=True,
    text=True,
)

print("STDOUT:")
print(res.stdout)

print("STDERR:")
print(res.stderr)

if res.returncode == 0:
    print("LoQI generation completed.")
else:
    print(f"LoQI generation failed with return code {res.returncode}.")
```
