## Generating LoQI Conformations for CN and NN Ligands

```python
from pathlib import Path
import subprocess
import sys
import os

# Project paths
PROJECT_ROOT = Path("/Volumes/T7/ML/IR_Ligand_ML")

generate_loqi_dir = Path(PROJECT_ROOT, "Generate_LoQI")
script_path = generate_loqi_dir / "generate_loqi_generic.py"

# LoQI repo paths
loqi_root = generate_loqi_dir / "LoQI"

# Input/output paths
cn_csv = Path(PROJECT_ROOT, "input", "ML-CN-Smiles.csv")
nn_csv = Path(PROJECT_ROOT, "input", "ML-NN-Smiles.csv")
output_base = Path(PROJECT_ROOT, "Generate_LoQI", "output")

cmd = [
    sys.executable, str(script_path),
    "--csvs", str(cn_csv), str(nn_csv),
    "--output-base", str(output_base),
    "--n-conformers", "12",
    "--batch-size", "50",
    "--sample-script", str(loqi_root / "scripts" / "sample_conformers.py"),
    "--config", str(loqi_root / "scripts" / "conf" / "loqi" / "loqi.yaml"),
    "--ckpt", str(loqi_root / "data" / "loqi.ckpt"),
]

env = os.environ.copy()
env["KMP_DUPLICATE_LIB_OK"] = "TRUE"

print("Running:")
print(" ".join(cmd))

res = subprocess.run(
    cmd,
    cwd=str(generate_loqi_dir),
    env=env,
    capture_output=True,
    text=True,
)

print("STDOUT:")
print(res.stdout)

print("STDERR:")
print(res.stderr)

if res.returncode == 0:
    print("LoQI generation completed.")
else:
    print(f"LoQI generation failed with return code {res.returncode}.")
```
